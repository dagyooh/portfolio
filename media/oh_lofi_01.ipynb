```python
import random
from fractions import Fraction

from klotho import plot, play
from klotho.tonos import Tonnetz, Pitch
from klotho.topos.shapes import Shape, translate, normalize

# helper functions for later
def kind(shape):
    """'major' for up-pointing triangles, 'minor' for down-pointing."""
    return 'major' if normalize(shape) == UP else 'minor'

def spell(shape):
    """Chord name from a triad: root pitch class + Maj/min."""
    root = Pitch.from_freq(tz.reference_pitch.freq * float(tz.get_ratio(min(shape))))
    return root.pitchclass + ('Maj' if kind(shape) == 'major' else 'min')
```

```python
def make_harmony(tour, span=1, tempus=TEMPUS, beat=BEAT, bpm=BPM, inst='hazePad', amp=0.75, **kwargs):
    seq = UTS()
    amp_pat = Pattern([amp, amp * 0.8])

    for shape in tour:
        # true, unreduced frequency per vertex -- matches the lattice plot's height
        freqs = tuple(
            tz.reference_pitch.freq * float(tz.get_ratio(v))
            for v in shape
        )
        uc = UC(span=span, tempus=tempus, prolatio='d', beat=beat, bpm=bpm, inst=ens.voice(inst))
        uc.root.set(freq=freqs, amp=next(amp_pat), **kwargs)
        seq.append(uc)

    return seq

harmony1 = make_harmony(tour)
play(harmony1.repeat(2))


# shorter chord times (span=1) and a shaped envelope
#play(make_harmony(chords_from_tour, inst=ens['hazePad'], amp=0.7, attackTime=0.1, decayTime=1.5, sustainLevel=0.1, releaseTime=0.5))
```

```python
def make_melody(notes, span=1, tempus=TEMPUS, beat=BEAT, bpm=BPM, inst='fd_blip', amp=0.8, octave_shift=0, prolatio='d', **kwargs):
    seq = UTS()
    for vertex in notes:
        freq = tz.reference_pitch.freq * float(tz.get_ratio(vertex)) * (2 ** octave_shift)
        uc = UC(span=span, tempus=tempus, prolatio=prolatio, beat=beat, bpm=bpm, inst=ens.voice(inst))
        uc.root.set(freq=freq, amp=amp, **kwargs)
        seq.append(uc)
    return seq
```

```python
def make_lattice_rhythm(notes, prolatio, span=1, tempus=TEMPUS, beat=BEAT, bpm=BPM,
                         inst='fd_bell', amp=0.8, octave_shift=0, **kwargs):
    seq = UTS()
    uc = UC(span=span, tempus=tempus, prolatio=prolatio, beat=beat, bpm=bpm, inst=inst)

    freqs = [
        tz.reference_pitch.freq * float(tz.get_ratio(v)) * (2 ** octave_shift)
        for v in notes
    ]
    uc.leaves.set(freq=Pattern(freqs), amp=amp, **kwargs)

    for node, dur in zip(uc.leaves, uc.durations):
        if dur < 0:
            uc.make_rest(node)

    seq.append(uc)
    return seq
```

```python
def make_lattice_rhythm(notes, prolatio, span=1, tempus=TEMPUS, beat=BEAT, bpm=BPM,
                         inst='fd_bell', amp=0.8, octave_shift=0, **kwargs):
    seq = UTS()
    uc = UC(span=span, tempus=tempus, prolatio=prolatio, beat=beat, bpm=bpm, inst=inst)

    freqs = [
        tz.reference_pitch.freq * float(tz.get_ratio(v)) * (2 ** octave_shift)
        for v in notes
    ]
    uc.leaves.set(freq=Pattern(freqs), amp=amp, **kwargs)

    for node, dur in zip(uc.leaves, uc.durations):
        if dur < 0:
            uc.make_rest(node)

    seq.append(uc)
    return seq


def n_sounding_slots(prolatio):
    """How many non-rest slots a flat prolatio tuple has -- your notes list should match this length."""
    return sum(1 for w in prolatio if w > 0)
```
